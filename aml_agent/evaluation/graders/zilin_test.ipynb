```python
#modify the following function to use LLM judge to evaluate if the aggregated result is correct or not, and return a score of 0 or 1 with a reason.     
from typing import Any
from aml_agent.evaluation.types import Evaluation
from aml_agent.evaluation.graders.llm_judge import (
    LLMJudgeConfig,
    run_llm_judge_structured,
    build_judge_error_evaluation,
)

SYSTEM_PROMPT = "You are an expert with numbers. Return JSON: {\"score\": 0 or 1, \"reason\": \"…\"}"
USER_INSTRUCTIONS = """You are evaluating an aggregated computation.
Return JSON: {\"score\": 0 or 1, \"reason\": \"...\"}.
score=1 if output is correct vs expected_output, score=0 otherwise."""  

def _build_expected_aggregation_text(expected_df):
    if expected_df is None or expected_df.empty:
        return "A total of 0 transactions with a cumulative amount of 0.00 were identified."

    import pandas as _pd
    if not isinstance(expected_df, _pd.DataFrame):
        expected_df = _pd.DataFrame(expected_df)

    expected_df = expected_df.copy()
    if 'amount' in expected_df.columns:
        expected_df['amount'] = expected_df['amount'].astype(float)
        total_amount = expected_df['amount'].sum()
    else:
        total_amount = 0.0

    total_tx = len(expected_df)

    counterparties = set()
    if 'sender_name' in expected_df.columns and 'receiver_name' in expected_df.columns:
        counterparties.update(expected_df['sender_name'].dropna().astype(str).tolist())
        counterparties.update(expected_df['receiver_name'].dropna().astype(str).tolist())
    num_counterparties = len(counterparties)

    return (
        f"A total of {total_tx} transactions with a cumulative amount of {total_amount:,.2f} "
        f"were identified with {num_counterparties} distinct counterparties."
    )


def LLM_judge_aggregation(
    input: Any,
    output: Any,
    expected_output: Any,
    metadata: dict[str, Any] | None = None,
    **kwargs: Any,
) -> list[Evaluation]:
    try:
        if hasattr(expected_output, 'to_dict') or isinstance(expected_output, dict):
            import pandas as _pd
            expected_df = _pd.DataFrame(expected_output)
        elif 'pandas' in str(type(expected_output)).lower():
            expected_df = expected_output
        else:
            # In case expected_output is a scalar or already summary text, don't treat as DataFrame
            expected_df = None

        expected_aggregation_text = (
            _build_expected_aggregation_text(expected_df)
            if expected_df is not None
            else str(expected_output)
        )

        user_prompt = (
            f"{USER_INSTRUCTIONS}\n\n"
            f"Observed summary text:\n{output}\n\n"
            f"Expected aggregation text:\n{expected_aggregation_text}\n\n"
            f"Please compare the observed summary text with expected aggregation text and return JSON with score 0 or 1 and reason." 
        )

        result = run_llm_judge_structured(
            metric_name="llm_aggregation_score",
            system_prompt=SYSTEM_PROMPT,
            user_prompt=user_prompt,
            config=LLMJudgeConfig(temperature=0.0),
        )

        raw_score = result.get("score", 0)
        score = 0
        if isinstance(raw_score, (int, float)):
            score = 1 if int(raw_score) == 1 else 0
        else:
            try:
                score = 1 if int(str(raw_score).strip().split(".")[0]) == 1 else 0
            except Exception:
                score = 0

        reason = result.get("reason", "No reason provided")

        return [Evaluation(name="llm_aggregation_score", value=score, comment=reason)]
    except Exception as exc:
        # pass exceptions into judge error evaluation (same behavior as initial implementation)
        return [build_judge_error_evaluation(metric_name="llm_aggregation_score", error=exc)]
```

```python
from typing import Any
import json
from aml_agent.evaluation.types import Evaluation

def sql_result_grader(
    output_path: Any,
    **kwargs: Any,
) -> list[Evaluation]:
    with open(output_path, 'r') as f:
        artifacts = json.load(f)

    try:
        input_name = artifacts['tool_calls'][0]['args']['keyword']
    except:
        input_name = artifacts['report']['subject']
    input_name = input_name.upper() #convert to uppercase for case-insensitive comparison

    
    #remove punctuation from input_name
    input_name = ''.join(e for e in input_name if e.isalnum() or e.isspace())

    data = pd.read_csv('data/transactions.csv')
    expected_df = data[(data.sender_name == input_name) | (data.receiver_name == input_name)] #expected output dataframe

    if not len(artifacts['sql_results']):
        if not len(expected_df):
            return[Evaluation(name="sql_result_score", value=1, comment="No sql result found in artifacts, which is correct since no matching transactions found in data")]
        else:
            observed_aggregation_text = artifacts.get('report_markdown', '')
            if observed_aggregation_text:
                llm_eval = LLM_judge_aggregation(
                    input=input_name,
                    output=observed_aggregation_text,
                    expected_output=expected_df,
                )
                if llm_eval:
                    return llm_eval
        return [Evaluation(name="sql_result_score", value=0, comment="0 if no sql result found in artifacts")]
    
    try:
        output = artifacts['sql_results'][0]['rows']
        output_df = pd.DataFrame(output, columns=artifacts['sql_results'][0]['columns']) #actual output dataframe

        expected_df = expected_df[artifacts['sql_results'][0]['columns']]

        if output_df.shape!= expected_df.shape:
            score = 0
        else:
            expected_df_sorted = expected_df.sort_values(by="transaction_id").reset_index(drop=True)
            output_df_sorted = output_df.sort_values(by="transaction_id").reset_index(drop=True)
            expected_df_sorted['amount'] = expected_df_sorted['amount'].astype(float)
            output_df_sorted['amount'] = output_df_sorted['amount'].astype(float)
            if expected_df_sorted.equals(output_df_sorted):
                score = 1
            else:
                score = 0.5
        #return output_df,expected_df
    except:
        return [Evaluation(name="sql_result_score", value=0, comment="0 if sql result is not in expected format")]
    return [Evaluation(name="sql_result_score", value=score, comment="1 if sql result matches expected output, 0.5 if shape matches but content does not match, 0 if shape does not match")]
```

```python
#run sql_result_grader on all test cases in output/test_cases
import os   
test_cases = [f for f in os.listdir('output/test_cases/test_run') if f.endswith('.artifacts.json')]
for test_case in test_cases:
    print(f"Grading {test_case}...")
    result = sql_result_grader(f'output/test_cases/test_run/{test_case}')
    print(result)
```
